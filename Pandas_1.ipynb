```python
import pandas as pd
```

```python
# Q. What is difference between Series and Dataframe?

# Series -(rows)--collection of elements --1D
# DataFrame --(rows,columns) --2D

# Series --similar to single column
# DataFrame --similar to an Excel table

# Series --Has an index
# DataFrame--has row and column indexes

# series --can contain different data type
# DataFrame --different columns can have different data types
```

```python
# To select rows and columns we have 2 options
# 1. loc: label based selection
# 2. iloc: index based selection
```

```python
# Q. What is loc and iloc

# Loc -Selects data using label/names
# it is label based
# Uses row/column names
# df.loc[row_label,column_label]

# iloc--Selects data using integer position
# position based
# uses row/column numbers
# df.iloc[row_position,column_position]
```

```python
# Q. iloc vs loc

# loc -- label based selection
# iloc --integer position based selection

# loc --uses rows/column labels
# iloc --uses row/column index position

# loc - df.loc[1,'name']
# iloc - df.iloc[1,0]

# loc - can use column names
# iloc --uses only integer position

# loc - endpoints in slices are included
# iloc - endpoint in slices is excluded
```
