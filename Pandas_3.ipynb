```python
# 1. append()
```

```python
# DataFrame.append() is used in Pandas to add rows from another DataFrame,
# Series,dictionary or list to the end of an existing DataFrame
# It returns a new DataFrame and do not modify the original one
```

# As of Pandas 2.0 DataFrame.append() has been removed.
# The recommended alternative is pd.concat()

```python
# This example appends one DataFrame to another
import pandas as pd

a = pd.DataFrame({'a': [1, 2, 3], 'b': [4, 5, 6]})
b = pd.DataFrame({'a': [7, 8, 9], 'b': [10, 11, 12]})

result = a.append(b)
print(result)
```

```python
# output ---
#    a  b
# 0  1  4
# 1  2  5
# 2  3  6
# 0  7  10
# 1  8  11
# 2  9  12
```

```python
# 2. concatenate()
```

```python
# You can combine two or more Pandas Object (DataFrame or Series) using the
# pd.concat() function
# Syntax for concatenation:
# pd.concat(objs, axis =0,join='outer',ignore_index=False, keys=None, verify_integrity=False)

# objs: A sequence or list of series or DataFrame objects to combine
# axis : The axis to concatenate along( 0 for row-wise/vertically, 1 for column-wise/horizontally)
# join: How to handle indexes on other oxes('outer' for the union,'inner' for the intersection)
# ignore_index : If True, discards the original index values and assigns a clean integer index(0,1,2,...)
# keys : Create a hierarchical index using passed keys

import pandas as pd
df1= pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df2= pd.DataFrame({'A': [7, 8, 9], 'B': [10, 11, 12]})


result = pd.concat([df1, df2])
print(result)
```

```python
# pd.concat([df1,df2]) stacks DataDrames row-wise and keeps their original indices, causing duplicates.
```

```python
# 3. merge()
```

```python
# The pd.merge() function in Pandas combines two DataFrames
# using database-style joins based on shared columns or indices

import pandas as pd

data1 = {
  "name": ["Sally", "Mary", "John"],
  "age": [50, 40, 30]
}

data2 = {
  "name": ["Sally", "Peter", "Micky"],
  "age": [77, 44, 22]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

newdf = df1.merge(df2,how='right')
print(newdf)
```

```python
import pandas as pd

data1 = {
  "name": ["Sally", "Mary", "John"],
  "age": [50, 40, 30]
}

data2 = {
  "name": ["Sally", "Peter", "Micky"],
  "age": [77, 44, 22]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

newdf = df1.merge(df2,how='left')
print(newdf)
```

```python
# All parameters except right, are keyword arguments.
```

```python
# right	 	  Required. A DataFrame, a Series to merge with
```

```python
# how ---'left','right','outer','inner','cross'
# Optional. Default 'inner' Specifies how to merge
```

```python
# on --string list --Optional. Specifies in what level to do the merging
```

```python
#left_on--string lsit--Optional. Specifies in what level to do the merging on the DataFrame to the left
```

```python
# right_on-string list --Optional. Specifies in what level to do the merging on the DataFrame to the right
```

```python
# left_index --True False --Optional. Default False. Whether to use the index from the left DataFrame as join key or not
```

```python
# right_index --True False--Optional. Default False. Whether to use the index from the right DataFrame as join key or not
```

```python
# Sort --True False --Optional. Default False. Specifies whether to sort the DataFrame by the join key or not
```

```python
# suffixes --List--Optional. Default '_x', '_y''. Specifies a list of strings to add for overlapping columns
```

```python
# copy -- True False -- Optional. Default True. Specifies whether to keep copies or not
```

```python
# 4. pivot()

# the df.pivot() function in Pandas reshapes a DataFrame from a long
# format to a wide format using unique values for specified index and column axes
```

```python
# Syntax:
# df.pivot(index=None, columns=None, values=None)
# index: Specifies the column label for the new row index(defaults to the existing index if omitted)
# columns: Specifies the column label to create the new columns
# values : Specifies the column(s) for populationg values; if omitted, all remaining
# columns from a hierarchical multi-index
```

```python
import pandas as pd
data = {
    'Date': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02'],
    'City': ['New York', 'Los Angeles', 'New York', 'Los Angeles'],
    'Temperature': [32, 75, 30, 77]
}
df =pd.DataFrame(data)
pivot_df = df.pivot(index ='Date',columns ='City',values ='Temperature')
print(pivot_df)
```

```python
# 5. join()
# .join() : Combines columns from another DataFrame or Series primarily
# based on indices rather than columns
```

```python
# Types of Joins (how parameter)

# Inner Join (how='inner'):
#Returns only the rows where matching keys exist in both DataFrames.

#Left Join (how='left'): Returns all rows from the left DataFrame and the matching rows from the right DataFrame (fills missing with NaN).

# Right Join (how='right'): Returns all rows from the right DataFrame and the matching rows from the left DataFrame.

# Outer Join (how='outer'): Returns all rows from both DataFrames, filling missing matches with NaN.

# Cross Join (how='cross'): Creates the Cartesian product (every combination of rows) from both DataFrames.
```

```python
df1 = pd.DataFrame({'name': ['Alice', 'Bob']}, index=[1, 2])
df2 = pd.DataFrame({'score': [85, 90]}, index=[1, 2])

# Join df2 onto df1 using index
result = df1.join(df2)
print(result)
```
